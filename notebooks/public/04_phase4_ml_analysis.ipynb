```python
import pandas as pd,os,numpy as np

root="astronomy_exp005_backup/results"
src=os.path.join(root,"exp035_source_feature_inventory.csv")
f=pd.read_csv(src)

print("="*62)
print("EXP-036 — ML-READY SOURCE MATRIX")
print("="*62)

# Keep source identity/provenance outside the ML feature matrix
id_cols=[
    "source_id","group_id","ra","dec",
    "validator_positive","validator_status",
    "validator_w1_chi2","validator_w1_snr","validator_quality",
    "old_label","baseline_positive","old_positive","agreement",
    "loo_status","image_status","external_status","known_class_status"
]

# Columns that directly encode validation outcomes or provenance
leak_cols=[
    "validator_positive","validator_status",
    "old_label","baseline_positive","old_positive","agreement",
    "validator_w1_chi2","validator_w1_snr","validator_quality"
]

# Candidate numerical feature columns
feature_cols=[]
for c in f.columns:
    if c in id_cols or c in leak_cols:
        continue
    if pd.api.types.is_numeric_dtype(f[c]):
        feature_cols.append(c)

X=f[feature_cols].copy()

# Remove columns with no information
constant=[c for c in X.columns if X[c].nunique(dropna=True)<=1]
X=X.drop(columns=constant)

# Keep missingness explicit with separate indicators.
# Do NOT impute yet; EXP-037 will decide the appropriate strategy.
missing_cols=[c for c in X.columns if X[c].isna().any()]

M=pd.DataFrame(index=f.index)
for c in missing_cols:
    M[c+"_available"]=f[c].notna().astype(int)

Xml=pd.concat([X,M],axis=1)

# Metadata for every ML row
meta=f[["source_id","group_id","ra","dec"]].copy()

# Save
x_path=os.path.join(root,"exp036_ml_features.csv")
m_path=os.path.join(root,"exp036_ml_metadata.csv")
Xml.to_csv(x_path,index=False)
meta.to_csv(m_path,index=False)

print("[OK] Sources:",len(f))
print("[OK] Candidate numeric features:",len(feature_cols))
print("[OK] Constant features removed:",len(constant))
print("[OK] ML features:",X.shape[1])
print("[OK] Availability indicators:",len(missing_cols))
print("[OK] Final matrix columns:",Xml.shape[1])
print("[OK] Final matrix rows:",len(Xml))

print("\nExcluded validation/leakage fields:")
for c in leak_cols:
    if c in f.columns:
        print(" -",c)

print("\nMissing-value features:")
for c in missing_cols:
    print(f" - {c}: {f[c].isna().sum()} missing")

print("\nFinal ML matrix:")
print(Xml.head().to_string())

print("\n[OK] Saved:",x_path)
print("[OK] Saved:",m_path)
print("\nSTATUS: EXP-036 COMPLETE")
```

```python
import pandas as pd,os,numpy as np

root="astronomy_exp005_backup/results"
xp=os.path.join(root,"exp036_ml_features.csv")
mp=os.path.join(root,"exp036_ml_metadata.csv")

X=pd.read_csv(xp)
meta=pd.read_csv(mp)

print("="*62)
print("EXP-037 — ML MATRIX INTEGRITY AUDIT")
print("="*62)

print("[OK] Rows:",len(X))
print("[OK] Columns:",len(X.columns))

# 1. Constant / near-constant features
num=X.select_dtypes(include=np.number)

constant=[]
near_constant=[]
for c in num.columns:
    vc=num[c].value_counts(normalize=True,dropna=False)
    if len(vc)<=1:
        constant.append(c)
    elif vc.iloc[0]>=0.99:
        near_constant.append(c)

print("\nVariance:")
print("constant:",len(constant))
print("near-constant:",len(near_constant))
if near_constant:
    print("near-constant features:",near_constant)

# 2. Duplicate columns
duplicates=[]
cols=list(X.columns)
for i in range(len(cols)):
    for j in range(i+1,len(cols)):
        a,b=cols[i],cols[j]
        if X[a].equals(X[b]):
            duplicates.append((a,b))

print("\nExact duplicate features:",len(duplicates))
for a,b in duplicates:
    print(" -",a,"==",b)

# 3. Correlation structure
corr=num.corr(min_periods=20)
pairs=[]
for i in range(len(corr.columns)):
    for j in range(i+1,len(corr.columns)):
        a,b=corr.columns[i],corr.columns[j]
        r=corr.iloc[i,j]
        if pd.notna(r) and abs(r)>=0.95:
            pairs.append((a,b,float(r)))

print("\nHighly correlated pairs |r| >= 0.95:",len(pairs))
for a,b,r in sorted(pairs,key=lambda x:-abs(x[2])):
    print(f" - {a} <-> {b}: r={r:.4f}")

# 4. Missingness
print("\nMissingness:")
for c in X.columns:
    n=int(X[c].isna().sum())
    if n:
        print(f" - {c}: {n}/{len(X)} ({100*n/len(X):.1f}%)")

# 5. Availability indicators
avail=[c for c in X.columns if c.endswith("_available")]
print("\nAvailability indicators:",len(avail))
for c in avail:
    print(f" - {c}: {X[c].value_counts(dropna=False).to_dict()}")

# 6. Scale summary
summary=pd.DataFrame({
    "feature":num.columns,
    "min":[num[c].min() for c in num.columns],
    "median":[num[c].median() for c in num.columns],
    "max":[num[c].max() for c in num.columns],
    "std":[num[c].std() for c in num.columns],
    "missing":[num[c].isna().sum() for c in num.columns]
})
summary.to_csv(os.path.join(root,"exp037_feature_scale_summary.csv"),index=False)

# 7. Leakage-name audit
leak_terms=[
    "validator","label","positive","baseline","old_","agreement",
    "class","target","verdict","candidate"
]
suspect=[c for c in X.columns if any(t in c.lower() for t in leak_terms)]

print("\nPotential leakage-name matches:")
print(suspect if suspect else "None detected")

# 8. Metadata integrity
print("\nMetadata:")
print("metadata rows:",len(meta))
print("metadata unique IDs:",meta["source_id"].nunique())
print("feature rows:",len(X))

status="PASS"
if len(X)!=len(meta) or meta["source_id"].nunique()!=len(meta):
    status="FAIL"

print("\n[OK] Scale summary saved.")
print("\nSTATUS:",status)
```

```python
import pandas as pd
import numpy as np
import os

root="astronomy_exp005_backup/results"
inp=os.path.join(root,"exp036_ml_features.csv")
out=os.path.join(root,"exp038A_canonical_features.csv")
meta_in=os.path.join(root,"exp036_ml_metadata.csv")
meta_out=os.path.join(root,"exp038A_canonical_metadata.csv")

X=pd.read_csv(inp)
meta=pd.read_csv(meta_in)

print("="*62)
print("EXP-038A-R — CORRECTED CANONICAL FEATURE MATRIX")
print("="*62)

value_features=[
    "n_observations",
    "w1_weighted_mean",
    "w1_std",
    "w1_median",
    "w1_range",
    "w1_reduced_chi2",
    "w1_mad_sigma",
    "w2_mean",
    "w2_std",
    "median_w1_snr",
    "time_span_days",
    "median_position_offset_arcsec",
    "max_position_offset_arcsec",
    "good_quality_fraction",
    "flagged_count",
    "flux_ratio",
    "influential_median_snr",
    "normal_median_snr",
    "loo_min_chi2",
    "loo_max_chi2",
    "loo_fraction_ge2",
    "loo_n_observations_tested"
]

# One indicator per genuinely distinct optional feature group.
# The underlying availability patterns are identical within each group.
availability_features={
    "image_features_available":"flux_ratio_available",
    "loo_features_available":"loo_min_chi2_available"
}

missing=[c for c in value_features if c not in X.columns]
if missing:
    raise ValueError(f"Missing value features: {missing}")

C=X[value_features].copy()

for new_col,source_col in availability_features.items():
    if source_col not in X.columns:
        raise ValueError(f"Missing availability source: {source_col}")
    C[new_col]=X[source_col].astype("int8")

# Check for exact duplicates.
duplicates=[]
cols=list(C.columns)

for i in range(len(cols)):
    for j in range(i+1,len(cols)):
        if C[cols[i]].equals(C[cols[j]]):
            duplicates.append((cols[i],cols[j]))

print("[OK] Original columns:",len(X.columns))
print("[OK] Value features:",len(value_features))
print("[OK] Availability indicators:",len(availability_features))
print("[OK] Final columns:",len(C.columns))
print("[OK] Rows:",len(C))
print("[OK] Exact duplicate columns remaining:",len(duplicates))

if duplicates:
    for a,b in duplicates:
        print(" -",a,"==",b)

print("\nMissingness:")
for c in C.columns:
    n=int(C[c].isna().sum())
    if n:
        print(f" - {c}: {n}/{len(C)} ({100*n/len(C):.1f}%)")

print("\nAvailability:")
for c in availability_features:
    print(f" - {c}:",C[c].value_counts(dropna=False).to_dict())

print("\nFeature groups:")
print(" temporal:")
print("  n_observations, time_span_days, w1_std, w1_range,")
print("  w1_reduced_chi2, w1_mad_sigma")

print(" photometric:")
print("  w1_weighted_mean, w1_median, w2_mean, w2_std,")
print("  median_w1_snr, flux_ratio")

print(" systematic:")
print("  good_quality_fraction, flagged_count,")
print("  median_position_offset_arcsec, max_position_offset_arcsec")

print(" robustness:")
print("  loo_min_chi2, loo_max_chi2,")
print("  loo_fraction_ge2, loo_n_observations_tested")

print(" image:")
print("  influential_median_snr, normal_median_snr")

status="PASS"

if len(C)!=len(X):
    status="FAIL"

if len(meta)!=len(C):
    status="FAIL"

if meta["source_id"].nunique()!=len(meta):
    status="FAIL"

if duplicates:
    status="FAIL"

if C.isna().all().any():
    status="FAIL"

C.to_csv(out,index=False)
meta.to_csv(meta_out,index=False)

print("\n[OK] Saved:",out)
print("[OK] Saved:",meta_out)
print("\nSTATUS:",status)
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.ensemble import IsolationForest
from sklearn.covariance import MinCovDet
from sklearn.metrics import pairwise_distances

root="astronomy_exp005_backup/results"
inp=os.path.join(root,"exp038A_canonical_features.csv")
meta_in=os.path.join(root,"exp038A_canonical_metadata.csv")
out=os.path.join(root,"exp038B_novelty_baseline.csv")

X=pd.read_csv(inp)
meta=pd.read_csv(meta_in)

print("="*62)
print("EXP-038B — FIRST UNSUPERVISED NOVELTY BASELINE")
print("="*62)

core_features=[
    "n_observations",
    "w1_weighted_mean",
    "w1_std",
    "w1_median",
    "w1_range",
    "w1_reduced_chi2",
    "w1_mad_sigma",
    "w2_mean",
    "w2_std",
    "median_w1_snr",
    "time_span_days",
    "median_position_offset_arcsec",
    "max_position_offset_arcsec",
    "good_quality_fraction",
    "flagged_count"
]

missing=[c for c in core_features if c not in X.columns]
if missing:
    raise ValueError(f"Missing core features: {missing}")

C=X[core_features].copy()

print("[OK] Sources:",len(C))
print("[OK] Core features:",len(core_features))
print("[OK] Missing values before preprocessing:",int(C.isna().sum().sum()))

imputer=SimpleImputer(strategy="median")
scaler=RobustScaler()

X_imp=imputer.fit_transform(C)
X_scaled=scaler.fit_transform(X_imp)

print("[OK] Median imputation applied")
print("[OK] Robust scaling applied")

# ------------------------------------------------------------
# Baseline 1: Isolation Forest
# ------------------------------------------------------------

iso=IsolationForest(
    n_estimators=500,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

iso.fit(X_scaled)

# Larger = more unusual.
iso_score=-iso.score_samples(X_scaled)

# ------------------------------------------------------------
# Baseline 2: Robust covariance distance
# ------------------------------------------------------------

# Minimum covariance determinant gives a robust estimate of
# the central population and covariance structure.
robust_cov=MinCovDet(
    random_state=42,
    support_fraction=0.75
)

robust_cov.fit(X_scaled)

mahal=robust_cov.mahalanobis(X_scaled)

# ------------------------------------------------------------
# Rank independently
# ------------------------------------------------------------

result=meta[["source_id"]].copy()

result["isolation_forest_score"]=iso_score
result["robust_mahalanobis"]=mahal

result["if_rank"]=result["isolation_forest_score"].rank(
    ascending=False,
    method="min"
).astype(int)

result["mahal_rank"]=result["robust_mahalanobis"].rank(
    ascending=False,
    method="min"
).astype(int)

result=result.sort_values(
    ["isolation_forest_score","robust_mahalanobis"],
    ascending=[False,False]
).reset_index(drop=True)

result["combined_rank_mean"]=(
    result["if_rank"]+result["mahal_rank"]
)/2

result=result.sort_values(
    "combined_rank_mean"
).reset_index(drop=True)

# ------------------------------------------------------------
# Attach useful source-level values
# ------------------------------------------------------------

for c in [
    "n_observations",
    "w1_reduced_chi2",
    "w1_std",
    "w1_range",
    "median_w1_snr",
    "good_quality_fraction"
]:
    result[c]=X[c].values

# Known Phase III survivors are NOT used by the models.
# They are added only after scoring for interpretation.
phase3={
    233:"TIER_2",
    234:"CONTROL_COMPATIBLE",
    1974:"TIER_1",
    1976:"TIER_2",
    1978:"TIER_1",
    1979:"CONTROL_COMPATIBLE"
}

result["phase3_status"]=result["source_id"].map(phase3).fillna("NOT_PHASE3_SURVIVOR")

result.to_csv(out,index=False)

print("\nTop 20 novelty-ranked sources:")
print(
    result[
        [
            "source_id",
            "combined_rank_mean",
            "if_rank",
            "mahal_rank",
            "isolation_forest_score",
            "robust_mahalanobis",
            "w1_reduced_chi2",
            "median_w1_snr",
            "phase3_status"
        ]
    ].head(20).to_string(index=False)
)

print("\nPhase III survivor ranks:")
print(
    result[
        result["source_id"].isin(phase3.keys())
    ][
        [
            "source_id",
            "combined_rank_mean",
            "if_rank",
            "mahal_rank",
            "phase3_status"
        ]
    ].sort_values("combined_rank_mean").to_string(index=False)
)

print("\n[OK] Saved:",out)
print("\nSTATUS: PASS")
```

```python
import os
import numpy as np
import pandas as pd

root="astronomy_exp005_backup/results"
feat_path=os.path.join(root,"exp038A_canonical_features.csv")
score_path=os.path.join(root,"exp038B_novelty_baseline.csv")
out=os.path.join(root,"exp038C_novelty_profile_audit.csv")

X=pd.read_csv(feat_path)
scores=pd.read_csv(score_path)

print("="*62)
print("EXP-038C — NOVELTY PROFILE + FEATURE CONTRIBUTION AUDIT")
print("="*62)

core_features=[
    "n_observations",
    "w1_weighted_mean",
    "w1_std",
    "w1_median",
    "w1_range",
    "w1_reduced_chi2",
    "w1_mad_sigma",
    "w2_mean",
    "w2_std",
    "median_w1_snr",
    "time_span_days",
    "median_position_offset_arcsec",
    "max_position_offset_arcsec",
    "good_quality_fraction",
    "flagged_count"
]

missing=[c for c in core_features if c not in X.columns]
if missing:
    raise ValueError(f"Missing features: {missing}")

# Robust feature center and scale using median + MAD.
med=X[core_features].median()
mad=(X[core_features]-med).abs().median()

# Prevent division by zero.
mad=mad.replace(0,np.nan)

Z=(X[core_features]-med)/mad

# Use absolute robust deviation as feature-level extremeness.
absZ=Z.abs()

# Top sources from the independent ML baseline.
top_sources=scores.sort_values(
    "combined_rank_mean"
).head(20)["source_id"].tolist()

rows=[]

for sid in top_sources:
    idx=X.index[X["source_id"]==sid]

    if len(idx)!=1:
        continue

    i=idx[0]
    feature_scores=absZ.loc[i].dropna().sort_values(ascending=False)

    top3=feature_scores.head(3)

    rows.append({
        "source_id":sid,
        "combined_rank_mean":float(
            scores.loc[scores["source_id"]==sid,"combined_rank_mean"].iloc[0]
        ),
        "if_rank":int(
            scores.loc[scores["source_id"]==sid,"if_rank"].iloc[0]
        ),
        "mahal_rank":int(
            scores.loc[scores["source_id"]==sid,"mahal_rank"].iloc[0]
        ),
        "top_feature_1":top3.index[0] if len(top3)>0 else "",
        "top_feature_1_abs_robust_z":float(top3.iloc[0]) if len(top3)>0 else np.nan,
        "top_feature_2":top3.index[1] if len(top3)>1 else "",
        "top_feature_2_abs_robust_z":float(top3.iloc[1]) if len(top3)>1 else np.nan,
        "top_feature_3":top3.index[2] if len(top3)>2 else "",
        "top_feature_3_abs_robust_z":float(top3.iloc[2]) if len(top3)>2 else np.nan,
        "n_features_abs_robust_z_ge_3":int((feature_scores>=3).sum()),
        "n_features_abs_robust_z_ge_5":int((feature_scores>=5).sum()),
        "max_abs_robust_z":float(feature_scores.max()) if len(feature_scores) else np.nan
    })

audit=pd.DataFrame(rows)

# Attach important raw values for interpretation.
raw_cols=[
    "n_observations",
    "w1_weighted_mean",
    "w1_std",
    "w1_range",
    "w1_reduced_chi2",
    "w1_mad_sigma",
    "w2_mean",
    "w2_std",
    "median_w1_snr",
    "time_span_days",
    "good_quality_fraction",
    "flagged_count"
]

for c in raw_cols:
    audit[c]=audit["source_id"].map(
        X.set_index("source_id")[c]
    )

phase3={
    233:"TIER_2",
    234:"CONTROL_COMPATIBLE",
    1974:"TIER_1",
    1976:"TIER_2",
    1978:"TIER_1",
    1979:"CONTROL_COMPATIBLE"
}

audit["phase3_status"]=audit["source_id"].map(
    phase3
).fillna("NOT_PHASE3_SURVIVOR")

# ------------------------------------------------------------
# Feature-level global extremeness
# ------------------------------------------------------------

feature_summary=pd.DataFrame({
    "feature":core_features,
    "median":med.values,
    "MAD":mad.values,
    "max_abs_robust_z":absZ.max().values,
    "n_sources_abs_z_ge_3":[int((absZ[c]>=3).sum()) for c in core_features],
    "n_sources_abs_z_ge_5":[int((absZ[c]>=5).sum()) for c in core_features]
})

feature_summary=feature_summary.sort_values(
    "max_abs_robust_z",
    ascending=False
)

# ------------------------------------------------------------
# Model disagreement
# ------------------------------------------------------------

corr=np.corrcoef(
    scores["if_rank"],
    scores["mahal_rank"]
)[0,1]

print("[OK] Sources:",len(X))
print("[OK] Core features audited:",len(core_features))
print("[OK] Top sources profiled:",len(audit))
print(f"[OK] IF-rank vs Mahalanobis-rank correlation: {corr:.4f}")

print("\nTop 20 novelty profiles:")
print(
    audit[
        [
            "source_id",
            "combined_rank_mean",
            "if_rank",
            "mahal_rank",
            "top_feature_1",
            "top_feature_1_abs_robust_z",
            "top_feature_2",
            "top_feature_2_abs_robust_z",
            "top_feature_3",
            "top_feature_3_abs_robust_z",
            "n_features_abs_robust_z_ge_3",
            "phase3_status"
        ]
    ].to_string(index=False)
)

print("\nMost extreme features across the population:")
print(feature_summary.head(10).to_string(index=False))

print("\nPhase III survivor profiles:")
print(
    audit[
        audit["source_id"].isin(phase3.keys())
    ][
        [
            "source_id",
            "combined_rank_mean",
            "if_rank",
            "mahal_rank",
            "top_feature_1",
            "top_feature_1_abs_robust_z",
            "top_feature_2",
            "top_feature_2_abs_robust_z",
            "top_feature_3",
            "top_feature_3_abs_robust_z",
            "phase3_status"
        ]
    ].sort_values("combined_rank_mean").to_string(index=False)
)

audit.to_csv(out,index=False)
feature_summary.to_csv(
    os.path.join(root,"exp038C_feature_extremeness_summary.csv"),
    index=False
)

print("\n[OK] Saved:",out)
print("[OK] Saved:",os.path.join(root,"exp038C_feature_extremeness_summary.csv"))
print("\nSTATUS: PASS")
```
